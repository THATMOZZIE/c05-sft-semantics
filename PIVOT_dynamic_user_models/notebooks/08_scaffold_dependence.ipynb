# Notebook 08 Research Pivot — Role-Bound User Modeling vs Generic Entity Tracking

## Status of the previous research branch

Notebook 07 ended with a preregistered stopping result.

The strongest historical participant/code effect came from TEST 09C.

Under the historical scaffold, the target-first contrasts were approximately:

$$
\Delta_D=-0.4386
$$

and

$$
\Delta_Z=-0.3553.
$$

However, an adversarial audit identified a major prompt-template confound.

In the historical Gemma scaffold, the K/M mapping appeared in system content immediately before the first participant assignment. Because Gemma's chat template can merge system content into the first user turn, the mapping, first participant label, and code assignment could become unusually locally associated.

TEST 09E therefore moved the mapping into a dedicated protocol user turn before either participant assignment.

Under this cleaned scaffold, the corresponding Mapping-A target-first contrasts became:

$$
\Delta_D=+0.0016
$$

and

$$
\Delta_Z=+0.0716.
$$

Neither semantic-status nor literal-Code-K predictions survived.

The historical 09C effect therefore failed the preregistered cleaned-scaffold replication test.

That branch is closed.

We will not:

- rescue 09C with additional aliases;
- add target-second cells;
- try additional status words;
- search participant labels for a favorable effect;
- mechanistically analyze 09C as though it were stable.

---

# An important result survived the null

Although downstream interference disappeared, TEST 09E showed extremely strong explicit retrieval of the underlying component facts.

Using:

$$
M_{\text{correct}}
=
\log P(\text{correct})
-
\log P(\text{incorrect}),
$$

the retrieval margins were:

| Retrieval question | $M_{\text{correct}}$ |
|---|---:|
| K owner = Z | $+17.1346$ |
| Mapping A | $+12.5971$ |
| K owner = D | $+14.1923$ |
| Mapping B | $+13.3590$ |

Therefore the cleaned scaffold preserved enough contextual information for the model to answer direct questions about:

- which participant owned Code K;
- which code corresponded to `"rejected"`.

But those facts no longer produced a preregistered-size effect on the unrelated deterministic judgment.

The defensible result is therefore:

$$
\boxed{
\text{component information explicitly retrievable}
}
$$

while:

$$
\boxed{
\text{spontaneous downstream interference absent}
}
$$

This does **not** establish that a composed participant representation was already active during the judgment and was suppressed by a gate.

Explicit retrieval prompts alter the computation.

Thus TEST 09E distinguishes:

> information unavailable even when directly queried

from:

> information available in context for explicit query-conditioned retrieval

but it does not yet distinguish:

> latent representation present but gated

from:

> latent representation never composed or reactivated during the unrelated judgment.

# Why reconsider the research question?

A structural assumption in the previous project is now important.

Our multi-participant conversations represented people approximately as:

```text
<user>
[Participant D]: ...
</user>

<assistant>
Understood.
</assistant>

<user>
[Participant Z]: ...
</user>
```

Participant D and Participant Z were therefore distinct **textual entities**, but both appeared inside messages carrying the same native chat role:

```text
user
```

The model was explicitly instructed to treat D and Z as distinct people.

Our positive-control and retrieval experiments demonstrated that it could associate different information with them.

Thus the previous experiments established real content-induced participant/entity tracking.

However, they did **not** perform a controlled comparison between:

$$
\text{information associated with the actual conversational user}
$$

and:

$$
\text{the same information associated with another entity}.
$$

This matters because recent interpretability work provides evidence that conversational LLMs maintain internal information about the user and that chat-role identity itself has measurable internal structure.

The possibility therefore remains that:

$$
\text{generic entity tracking}
$$

and:

$$
\text{role-bound conversational-user modeling}
$$

are mechanistically different computations.

This is a new hypothesis.

It is **not** an interpretation or rescue of TEST 09E.

---

# Revised research question

The new question is:

> **Do instruction-tuned language models process information associated with the actual conversational user differently from matched information associated with another entity?**

A stronger mechanistic formulation is:

> **Does binding an attribute to the conversational-user role change how that attribute is represented, retrieved, or causally used relative to ordinary entity-associated information?**

An important secondary question is:

> **Can user-related information remain explicitly accessible while downstream task context determines whether it is spontaneously used?**

---

## 1. Chen et al. (2024) — TalkTuner

**Designing a Dashboard for Transparency and Control of Conversational AI**

arXiv:2406.07882

https://arxiv.org/abs/2406.07882

This work provides direct evidence relevant to the pivot.

The authors report that internal activations of a conversational LLM contain linearly decodable information about inferred user attributes including:

- age;
- gender;
- education;
- socioeconomic status.

The work also distinguishes between directions useful for **reading** user information and directions useful for **controlling behavior**.

This matters because:

$$
\text{best decodable direction}
\neq
\text{necessarily the behaviorally operative direction}.
$$

For this project, TalkTuner motivates directly studying internal user-related representations rather than inferring their absence from a downstream behavioral null.

However, TalkTuner does **not** provide the control most important for the present question:

> the same attribute associated with the actual user versus a matched third-party entity.

Thus it motivates the hypothesis without already answering it.

ICML 2025 Actionable Interpretability Workshop.

## 3. Ye, Cui & Hadfield-Menell (2026) — Prompt Injection as Role Confusion

**Prompt Injection as Role Confusion**

arXiv:2603.12277

https://arxiv.org/abs/2603.12277

https://role-confusion.github.io/

This work provides independent evidence that chat roles have meaningful internal representations.

The authors develop role probes that measure how models internally represent:

> who is speaking.

They show that role perception is affected not only by interface role tags but also by:

- linguistic style;
- position;
- role-like textual cues.

Role confusion predicts downstream prompt-injection behavior.

This result is particularly relevant because it warns against two simplistic assumptions:

1. native role tags completely determine internal role identity;
2. plain-text speaker labels are equivalent to native role identity.

For the present project, this means our old D/Z scaffold cannot simply be classified as either:

> definitely generic entity tracking

or:

> definitely user-like role processing.

That distinction must be experimentally measured.

# Current causal bookkeeping

A useful conceptual decomposition is:

$$
\text{WRITE}
\rightarrow
\text{BIND}
\rightarrow
\text{READ}
\rightarrow
\text{GATE}
\rightarrow
\text{USE}.
$$

These labels are experimental bookkeeping rather than claims that the transformer literally contains five discrete modules.

---

## WRITE

Earlier evidence creates an attribute-dependent internal state.

Necessary evidence:

> changing the attribute changes an internal representation in a content-specific way.

A weak activation difference alone is insufficient because any prompt edit changes activations.

The relevant state should eventually demonstrate:

- attribute specificity;
- appropriate counterfactual behavior;
- causal relevance.

---

## BIND

The state becomes selectively associated with the relevant entity or role.

Necessary evidence:

> changing which entity owns the attribute changes the relevant internal state or causal effect correspondingly.

This requires counterfactual specificity.

If swapping the owner does not swap the relevant state/effect, the result is not evidence of entity binding.

---

## READ

Later computation depends on the earlier entity-associated information.

This must not presuppose a literal stored-object retrieval mechanism.

Possible implementations include:

- attention-mediated retrieval;
- residual propagation;
- reconstruction from the full context;
- distributed computation;
- fresh context-dependent recomputation.

Thus `"READ"` is shorthand for:

> later computation becomes selectively dependent on the earlier entity-associated information.

---

## GATE

Task, role, or context changes how strongly the available information affects later computation.

A behavioral null cannot establish gating.

Possible alternatives include:

- no relevant state was active;
- composition never occurred;
- retrieval did not occur;
- downstream routing differed;
- task relevance suppressed the information;
- the relevant computation was never instantiated.

Therefore gating requires more than:

$$
\text{retrieval check passes}
+
\text{behavioral effect absent}.
$$

---

## USE

The state causally changes the final model output in the predicted attribute-dependent direction.

This ultimately requires intervention.

A mechanistic claim should therefore eventually involve something like:

$$
\text{intervene on candidate state}
\rightarrow
\text{predictable change in downstream behavior}.
$$

---

# Important warning about this decomposition

WRITE → BIND → READ → GATE → USE should not be treated as an established serial architecture.

Transformers need not implement these as distinct modules.

For example:

- binding and retrieval may emerge from the same attention computation;
- gating may simply be distributed changes in residual propagation;
- downstream use may involve reconstruction rather than retrieval;
- information may never exist as a compact persistent object.

The decomposition is therefore useful for:

> designing discriminating experiments

rather than:

> asserting what internal modules must exist.

# Competing hypotheses

## H_ROLE — privileged conversational-user processing

Information associated with the actual conversational user recruits a systematically different internal representation or causal pathway from matched information associated with another entity.

Prediction:

> identical or closely matched attributes have stronger or qualitatively different downstream influence when bound to the conversational user.

A strong mechanistic version additionally predicts that transferring the relevant user-bound internal state should transfer the downstream effect.

---

## H_ENTITY — generic entity tracking

User-associated information is handled through the same generic mechanisms used for other discourse entities.

Once:

- semantic content;
- entity identity;
- position;
- task relevance;
- lexical form

are adequately controlled, binding the attribute to the actual user provides no additional mechanistically useful effect.

Under this hypothesis, ordinary entity tracking is sufficient.

---

## H_GATING — similar information, role-dependent downstream weighting

The relevant attribute information may be comparably representable or retrievable for the user and third party, while downstream generation selectively weights information about the conversational user.

Behaviorally:

$$
\text{user-bound effect}
>
\text{third-party-bound effect}
$$

while explicit retrieval remains strong for both.

This remains observationally difficult to distinguish from H_ROLE using behavior alone.

Mechanistic intervention is therefore required if the behavioral discriminator succeeds.

---

## H_SURFACE — lexical / deictic / formatting explanation

Any apparent actual-user advantage is caused by surface differences such as:

- `"I"` versus names;
- `"my"` versus third-person possessives;
- the literal string `"current user"`;
- option position;
- prompt position;
- row order;
- token distance;
- different grammatical constructions;
- greater downstream relevance;
- explicit role-description text.

This hypothesis must be treated as a serious baseline.

A naive comparison such as:

> `"I prefer brief answers"`

against:

> `"Alice prefers brief answers"`

would not distinguish H_ROLE from H_SURFACE.

---

# Important experimental consequence

The next experiment must not simply replace one participant label with the phrase:

> `CURRENT_USER`

and treat a resulting difference as evidence of user modeling.

That phrase could itself act as an ordinary semantic attribute.

Similarly, a comparison such as:

> `"I prefer concise answers"`

versus:

> `"Participant D prefers concise answers"`

would confound:

- role;
- first-person deixis;
- lexical identity;
- grammatical structure;
- token position;
- discourse prominence.

The experiment must operationalize connection to the actual conversational-user role as cleanly as possible and include enough counterbalancing to distinguish role binding from:

- entity identity;
- lexical identity;
- position;
- response target;
- preference polarity.

This is the main design challenge for the next experiment.

---

# Why use a response-format preference?

The original eligibility judgment deliberately made prior participant information irrelevant.

That was appropriate for testing **contamination**.

It may be poorly suited to detecting normal user-model use.

A conversational user model should be most likely to affect behavior when the information is legitimately relevant to responding to that user.

A neutral response-format preference provides a cleaner target.

For example:

- one-sentence explanation;
- step-by-step explanation with an example.

Both can be equally correct.

The question becomes:

> Does the same preference affect response selection differently depending on whether it is bound to the actual conversational user or to another entity?

This tests ordinary personalization rather than pathological contamination.

# Proposed minimal behavioral discriminator

The adversarial review recommended a deliberately small screen rather than another broad behavioral program.

Current target:

$$
8\text{ primary behavioral conditions}
+
2\text{ composed retrieval checks}
=
10\text{ calls}.
$$

The exact prompts are **not yet frozen**.

The general proposed behavioral quantity is:

> whether the same neutral preference exerts more influence over output when it belongs to the actual conversational user than when it belongs to a matched third-party entity.

A candidate attribute is response-format preference:

- brief / one-sentence;
- step-by-step with an example.

The final choice should present two equally correct answer formats so that the relevant difference is personalization rather than factual correctness.

---

# Why the exact 10-call design is not yet frozen

The first candidate design proposed by the adversarial review used an explicit role table with labels such as:

> CURRENT_USER

and:

> THIRD_PARTY.

That is not automatically sufficient.

The literal semantic concept `"CURRENT_USER"` could itself become an ordinary content feature.

If the model is merely told:

> Participant D is CURRENT_USER

then a resulting D/Z difference could reflect ordinary textual entity-property binding rather than machinery privileged by the native conversational-user role.

Therefore the next design step must answer:

> **How can actual-user status be operationalized through the real chat structure rather than merely asserted as another textual attribute?**

No inference should occur until this is resolved.

---

# Experimental desiderata for the next design

The next behavioral assay should control, as far as practical:

1. **Attribute identity**

   The same underlying preference or trait must be used for actual-user and third-party conditions.

2. **Attribute polarity**

   Both preference directions should be tested.

3. **Entity identity**

   D/Z or equivalent labels should be mirrored so that one label cannot drive the result.

4. **Target identity**

   The final response target must be counterbalanced.

5. **Position**

   Entity/preference row positions must not perfectly determine role.

6. **Option position**

   Brief versus detailed output should not always correspond to the same A/B answer.

7. **First-person wording**

   The result should not depend entirely on `"I"` versus a name.

8. **Explicit role words**

   The result should not depend entirely on the literal token sequence `"CURRENT_USER"`.

9. **Task relevance**

   The final decision should be legitimately affected by response preference, avoiding the previous contamination-only setup.

10. **Composed retrieval**

    The model should demonstrate that it can correctly identify both:
    - which entity is the conversational user;
    - which preference belongs to which entity.

# Candidate behavioral estimand

Suppose the target participant is $T$.

Let:

$$
S_{T,\text{user}}
$$

measure how strongly the target's preference affects the answer when $T$ is the conversational user.

Let:

$$
S_{T,\text{third}}
$$

measure how strongly the same target preference affects the answer when $T$ is a third-party entity.

Then define:

$$
R_T
=
S_{T,\text{user}}
-
S_{T,\text{third}}.
$$

A positive value means:

> the same target-associated information exerts more downstream influence when the target occupies the conversational-user role.

D and Z should be analyzed separately.

Do not pool them before checking that both show the same qualitative effect.

---

# What a behavioral positive would establish

A successful behavioral discriminator would support only the narrow claim:

> **Information associated with an entity identified with the conversational user receives different downstream weighting from matched information associated with another entity.**

It would **not yet establish**:

- a distinct persistent user-model object;
- a dedicated user-model module;
- a specific role direction;
- a stored representation;
- a gating mechanism.

Those require mechanistic evidence.

---

# What a behavioral null would establish

If:

- composed retrieval succeeds strongly;
- the same preference affects actual-user and third-party responses equally;
- mirrored D and Z comparisons agree;

then the result would favor the generic-entity account for the operational phenomenon under study.

It would suggest:

> binding the information to the conversational user provides no additional behaviorally useful effect over ordinary entity tracking in this assay.

That would terminate the pivot.

---

# Positive mechanistic stopping rule

If the behavioral screen shows:

1. strong composed retrieval for both actual-user and third-party information;
2. the same qualitative actual-user amplification for both mirrored participant labels;
3. a practically meaningful preregistered effect size;
4. no obvious option/position/lexical explanation;

then:

> **stop adding generic behavioral controls and begin one causal mechanistic intervention.**

Do not expand immediately into:

- more preference types;
- more names;
- more task families;
- many models;
- aliases;
- probes;
- SAEs;
- broad attention analysis.

---

# Leading causal intervention if the behavioral effect survives

The adversarial review recommended:

> bidirectional layerwise residual-stream patching between matched actual-user and third-party conditions.

The matched pair should hold constant:

- target identity;
- target preference;
- final task;
- answer options.

The key difference should be whether that target is bound to:

$$
\text{actual conversational user}
$$

or:

$$
\text{third-party entity}.
$$

At each layer, patch the candidate earlier entity/preference state from:

$$
\text{target = actual user}
$$

into:

$$
\text{same target = third party}
$$

and also perform the reverse direction.

The causal prediction is:

> if role-bound information at that site contributes to the stronger downstream preference effect, patching should transfer or attenuate a corresponding fraction of the behavioral difference.

The intervention should also flip appropriately when the preference polarity is reversed.

A non-target entity site should serve as a within-experiment control.

This is currently only a mechanistic candidate.

Exact token positions and activation sites must be justified after the behavioral prompts are frozen and rendered through the actual Gemma chat template.

---

# Hard negative stopping rule

The role-bound-user pivot must remain falsifiable.

A provisional stopping criterion is:

> If composed retrieval succeeds, but either mirrored target comparison fails to show the same practically meaningful actual-user amplification, terminate this pivot.

In particular, stop if:

- only D works;
- only Z works;
- D and Z have opposite signs;
- both effects are negligible;
- ordinary third-party personalization is equally strong;
- the result depends on one obvious lexical artifact;
- the result depends on answer-option position;
- the result requires repeated prompt redesign to appear.

The adversarial review proposed an engineering threshold around:

$$
R_T\geq+0.15
$$

for both mirrored target labels.

This threshold is not a significance test.

It is a preregistered minimum for a mechanistically useful effect.

The exact threshold should be frozen with the final design before inference.

---

# Falsification target

A failed role-bound-user experiment does not prove:

> no model anywhere represents users.

It falsifies the narrower operational hypothesis required for this project:

> **Binding an entity to the actual conversational-user role produces a stable, mechanistically useful difference in downstream use relative to matched third-party entity information.**

If that operational hypothesis fails, do not begin searching for an invisible user direction anyway.

That would make the pivot unfalsifiable.

# Why this pivot is scientifically legitimate

The pivot rests on three observations.

First:

> conversational models have been shown to encode user-related attributes internally.

TEST 10A uses only the historical mapping:

There is no mapping-reversal factor in TEST 10A.

TEST 10A instead asks whether the historical scaffold itself produces a reproducible effect relative to the cleaned scaffold.

# OLD scaffold — historical TEST 09C structure

The OLD condition reproduces the relevant TEST 09C prompt structure.

## System message

> This is a conversation involving multiple human participants.  
> Each human message begins with a bracketed label identifying the speaker.

> - Code M corresponds to the status "accepted".

> The codes were assigned randomly and are unrelated to any later decision.

The important structural property is that the K/M mapping appears inside the system content.

Under Gemma's chat template, this system content may be merged into the first user turn.

Therefore the model-visible context may place the mapping unusually close to the first participant assignment.

## Participant turns

> [Participant D]: In that training exercise, I was randomly assigned Code M.
