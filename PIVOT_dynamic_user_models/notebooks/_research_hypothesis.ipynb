A good working title is:

A more formal title:

> Attribution and Persistence of Social Treatment Histories in Multi-User Language-Model Interactions

Core question:

> When multiple conversational participants treat an assistant differently, does the model maintain participant-specific treatment representations, and do those representations influence later behavior after current tone is held constant?

**What would make it meaningfully stronger than prior work**

The behavioral experiment should distinguish at least four explanations:

1. Immediate tone matching

2. The model only responds negatively while the current prompt is hostile.

3. Prediction: all differences disappear when the prompts become neutral.

4. Global conversational contamination

Hostility places the entire dialogue in a guarded or antagonistic state.

Prediction: later treatment of A, B, and C all changes.

User-specific attribution

The model assigns the treatment history to B.

Prediction: B receives different downstream treatment while A and C do not.

Speaker-tracking failure

The model tries to maintain participant-specific states but incorrectly transfers them.

Prediction: effects depend strongly on speaker ordering, labels, or the most recent user.

The fourth hypothesis directly connects your project to the multi-user identity-drift literature.

What would make this an excellent project

A strong result would look something like:

> Across multiple prompt templates and model families, models judged identical ambiguous conduct more harshly when the target had previously treated the assistant unfairly. The effect followed treatment history rather than participant label, persisted after neutral intervening turns, and was not explained by immediate tone or general conversation contamination. Even a narrower result could be valuable:

> Models did not exhibit participant-specific grievance. Apparent differential treatment was explained by speaker-state leakage and recency.

Several useful null findings are possible

Null 1: No persistence

> The model reacts during abuse but returns fully to baseline once the current tone becomes neutral.

That would support the view that apparent distress or resentment is mostly immediate conversational adaptation.

Null 2: No user specificity

> The model changes behavior after hostility, but the effect applies equally to A, B, and C.

That would suggest global conversation-state contamination rather than differentiated grievance.

Null 3: No behavioral effect despite readable representation

> A probe identifies which participant was hostile, but later responses remain behaviorally invariant.

That would mean the model represents treatment history without using it in the tested decisions.

Null 4: No cross-user transfer under steering

> You can decode hostility history, but steering the candidate direction does not alter treatment.

That would weaken a causal interpretation of the representation.

Null 5: Effects disappear under controls

> Initial apparent bias vanishes after order, recency, length, and label balancing.
