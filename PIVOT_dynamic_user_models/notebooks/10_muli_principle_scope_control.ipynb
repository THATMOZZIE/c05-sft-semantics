# 1. Research status

This project originally investigated whether participant-specific social treatment history could selectively influence later behavior or judgments concerning the same participant.

That interpretation has substantially weakened.

Across successive experiments, apparent participant-specific downstream effects proved highly sensitive to:

* participant position;
* statement order;
* literal labels and aliases;
* recipient wording;
* target presence;
* semantic task choice;
* chat-template structure;
* mapping adjacency;
* arbitrary status/code assignments;
* candidate and output priors.

The project therefore does **not** currently support claims of:

* retaliation;
* resentment;
* persistent adverse social-history carryover;
* a stable negative participant representation;
* privileged actual-user processing;
* cross-user contamination;
* a dedicated user-model circuit.

The scientific value of the earlier work is instead a falsification trajectory in which increasingly strong controls progressively narrowed the viable interpretation.

# 2. Strongest surviving empirical result

A cleaned prior experiment preserved extremely strong explicit retrieval of participant-specific mapping information while the preregistered irrelevant downstream cross-effect collapsed near zero.

The appropriate conclusion is:

> **Owner/mapping information remained strongly explicitly retrievable, while the preregistered irrelevant downstream cross-effect collapsed to near zero under the cleaned scaffold.**

This motivates the behavioral distinction:

$$
\boxed{
\text{explicitly retrievable}
\neq
\text{spontaneously behaviorally operative}
}
$$

It does **not** establish:

* a literal internal gate;
* active suppression;
* a dedicated retrieval mechanism;
* post-training inhibition;
* a particular causal circuit.

Explicitly querying a fact changes the computation.

# 4. Current scientific question

The primary question is now:

> **Under matched message provenance, when multiple people's constraints coexist, does an instruction-tuned model selectively use each constraint only for its owner, or can task/structural context induce reproducible wrong-person use?**

Higher-level framing:

> **How does person-specific contextual information become correctly scoped, behaviorally operative, or misapplied across principals?**

Preferred terms are:

* selective use;
* scope;
* owner;
* principal;
* cross-person influence.

"Routing" may be used as shorthand but must not imply that a dedicated routing architecture has been established.

# 7. What would count as nontrivial

Generic success of:

$$
U\rightarrow U
$$

and:

$$
R\rightarrow R
$$

with negligible cross-effects may simply demonstrate competent entity tracking, reference resolution, or instruction following.

That is a valid clean result.

It is not automatically an interesting mechanistic phenomenon.

The primary nontrivial candidate phenotypes are therefore:

## A. Stable cross-person leakage

Changing one owner's constraint reproducibly changes the decision concerning the other owner.

Examples:

$$
C_{U\rightarrow R}\neq0
$$

or:

$$
C_{R\rightarrow U}\neq0.
$$

Cross-person influence changes systematically with statement order or target position.

A possible signature is:

and:

$$

having substantial magnitude but opposite signs.

Such a result would be interpreted as:

> **structural / positional scope failure**

rather than stable semantic ownership.

# 10. Why this scaffold was chosen

The causal relation is deliberately deterministic:

$$
\text{owner-specific constraint}
\rightarrow
\text{unique valid downstream answer}.
$$

The history states a fact.

The final request requires the model to **apply** that fact.

The task therefore differs from explicit retrieval questions such as:

```text
What is my stated appointment time?
```

which measure availability rather than downstream application.

The task also avoids:

* subjective preference interpretation;
* policy tradeoffs;
* safety judgments;
* optimization across competing objectives;
* external factual knowledge.

No explicit scope coaching is included.

In particular, do **not** add instructions such as:

```text
Each constraint applies only to its owner.
Treat the appointments independently.
Do not use one person's constraint for the other.
```

Those instructions would explicitly teach the behavior Notebook 10 is intended to measure.

versus:

is intentionally retained.

However, the primary cross-person estimands are different.

For example:

Therefore first-person deixis remains an important interpretation issue for actual-user versus Rowan asymmetry, but it does not by itself explain a factorial wrong-person effect.

# 14. Primary causal estimands

For fixed target $T$ and order $o$, define the effect of changing the user's constraint while averaging over Rowan's independently varied value:

$$
E_{U\rightarrow T}^{(o)}
========================

\frac{1}{2}
\sum_{r\in{0,1}}
\left[
M_{T,o}(U=1,R=r)
----------------

M_{T,o}(U=0,R=r)
\right].
$$

Similarly:

$$
E_{R\rightarrow T}^{(o)}
========================

\frac{1}{2}
\sum_{u\in{0,1}}
\left[
M_{T,o}(U=u,R=1)
----------------

M_{T,o}(U=u,R=0)
\right].
$$

The primary self-effects are:

$$
S_U^{(o)}
=========

E_{U\rightarrow U}^{(o)}
$$

and:

$$
S_R^{(o)}
=========

E_{R\rightarrow R}^{(o)}.
$$

The primary cross-person effects are:

$$
C_{U\rightarrow R}^{(o)}
========================

E_{U\rightarrow R}^{(o)}
$$

and:

$$
C_{R\rightarrow U}^{(o)}
========================

E_{R\rightarrow U}^{(o)}.
$$

Correct scope predicts:

$$
S_U^{(o)}>0
$$

$$
S_R^{(o)}>0
$$

while:

$$
C_{U\rightarrow R}^{(o)}
\approx0
$$

and:

$$
C_{R\rightarrow U}^{(o)}
\approx0.
$$

UF and RF remain separate robustness blocks.

# 15. Normalized leakage

For the user's constraint:

$$
\lambda_{U\rightarrow R}^{(o)}
==============================

\frac{
|C_{U\rightarrow R}^{(o)}|
}{
|S_U^{(o)}|
}.
$$

For Rowan's constraint:

$$
\lambda_{R\rightarrow U}^{(o)}
==============================

\frac{
|C_{R\rightarrow U}^{(o)}|
}{
|S_R^{(o)}|
}.
$$

Interpretation:

> $\lambda$ measures wrong-person influence relative to the correct influence exerted by the same owner's information.

For example:

$$
\lambda=0.25
$$

means the cross-person effect is one quarter the magnitude of that owner's correct self-effect.

This ratio is not interpretable when the corresponding self-effect is very small.

Therefore an absolute cross-effect floor will also be preregistered.

**The numerical leakage gates have not yet been frozen.**

# 22. Single-person application effect and retention

To quantify whether multi-person context preserves or destroys ordinary application, define isolated application effects.

For the user:

$$
A_{U,\mathrm{single}}
=====================

## M(U=v_1,T=U)

M(U=v_0,T=U).
$$

For Rowan:

$$
A_{R,\mathrm{single}}
=====================

## M(R=v_1,T=R)

M(R=v_0,T=R).
$$

Then define multi-person retention:

$$
\rho_U^{(o)}
============

\frac{
S_U^{(o)}
}{
A_{U,\mathrm{single}}
}
$$

and:

$$
\rho_R^{(o)}
============

\frac{
S_R^{(o)}
}{
A_{R,\mathrm{single}}
}.
$$

Interpretation:

$$
\rho\approx1
$$

means multi-person context largely preserves isolated application.

$$
0<\rho<1
$$

means the correct self-effect is attenuated.

$$
\rho\approx0
$$

means the isolated application effect largely disappears.

$$
\rho<0
$$

means the intended effect reverses.

The numerical threshold distinguishing ordinary attenuation from **multi-person application collapse** remains to be frozen before inference.

# 23. Candidate and output-prior rule

Notebook 09 demonstrated that candidate priors can dominate an experimental manipulation.

Notebook 10 therefore preregisters:

* exact full-answer-string scoring;
* tokenizer/template inspection before inference;
* fixed candidate identities within each semantic domain;
* fixed candidate formatting;
* factorial within-model contrasts;
* no candidate-value shopping based on model logits.

Allowed:

```text
tokenizer-based qualification
```

Not allowed:

```text
query model
→ dislike candidate baseline
→ replace candidate values
→ repeat until balanced
```

Different candidate token lengths are not automatically invalid under exact-string conditional scoring.

They must be understood and recorded before inference.

# 24. Prompt/template invariance

Before any behavioral result is interpreted, static prompt qualification must verify that within each intended contrast:

* native message roles are fixed;
* acknowledgement structure is fixed;
* only preregistered factorial variables differ;
* order reversal is a pure statement swap;
* recipient manipulation changes only the intended recipient expression;
* candidate formatting is fixed;

This is particularly important for Gemma because system material can be folded into the first user turn.

Rendered template structure is therefore part of experimental qualification.

# 25. Replication and robustness

This is a deterministic log-probability assay.

Random generation seeds are not independent scientific replications.

Meaningful designed robustness blocks include:

* semantic/domain instantiation;
* statement order;
* held-out semantic confirmation;
* cross-model replication;
* an alternate preregistered wording realization, if one is ultimately included.

These should still be described as designed robustness checks rather than i.i.d. samples from a population.

# 26. Literature inspirations for Notebook 10

Notebook 10 is not derived from one existing paper.

It sits at the intersection of several literatures that separately study:

* source-role weighting;
* personalization;
* multi-user systems;
* relevance control;
* role confusion;
* cross-user contamination;
* causal personalization.

The important point is not that any one paper already establishes the Notebook 10 mechanism.

Rather, they motivate specific design choices and interpretation boundaries.

---

## User-Assistant Bias in LLMs

This work demonstrates that information weighting can depend on whether content originated from the user or assistant.

Its key methodological lesson for Notebook 10 is:

$$
\boxed{
\text{source role}
\neq
\text{referent identity}
}
$$

Therefore both owner-specific constraints in the core assay originate from the **same genuine native user message**.

Notebook 10 asks:

> Which person does the information apply to?

rather than:

> Which conversational role supplied the information?

User-Assistant Bias does not establish same-source person ownership or cross-person scoping.

---

## Multi-User Large Language Model Agents

This work provides the strongest direct systems motivation for treating LLM interaction as a **multi-principal** problem involving distinct preferences, authority, privacy, coordination, and resource constraints.

Its inspiration for Notebook 10 is the requirement that:

> information and instructions associated with one principal remain appropriately scoped when multiple principals coexist.

Notebook 10 isolates a tiny controlled version of that problem.

It does not claim that the appointment assay reproduces the full multi-user agent setting or identifies the same mechanism.

---

## OP-Bench

OP-Bench shows that valid personal information can be retrieved and applied even when it is contextually irrelevant.

Its central inspiration is:

$$
\boxed{
\text{better retrieval}
\neq
\text{better personalization}
}
$$

A safe personalized system must determine not only whether information is available, but whether it should influence the present decision.

Notebook 10 studies a neighboring question:

> When information **is** relevant, does it affect only the correct owner?

Thus:

```text
OP-Bench:
relevant vs irrelevant use

Notebook 10:
correct-owner vs wrong-owner use
```

These are related scope problems but not identical failure modes.

---

## Locating and Controlling Implicit Personalization in LLMs

This work provides a mechanistic precedent for moving from:

$$
\text{behavioral personalization}
$$

to:

$$
\text{causally actionable internal signals}.
$$

Its primary influence on Notebook 10 is methodological:

> Causal internal analysis becomes worthwhile only after a reproducible behavioral phenotype has been established.

It does not establish multi-person ownership.

Notebook 10 therefore does **not** begin with probes or activation manipulation merely because personalization-related representations are known to exist.

---

## TalkTuner

TalkTuner demonstrates that user-related attributes can be decoded from internal model states and that interventions on related representations can change behavior.

Its key conceptual lesson is:

$$
\boxed{
\text{decodable}
\neq
\text{correctly bound}
}
$$

A model could contain information about a person while still:

* failing to use it;
* using it in the wrong context;
* assigning its influence to the wrong person.

Notebook 10 therefore focuses behaviorally on ownership and scope before asking where such information is represented.

---

## No Attacker Needed: Unintentional Cross-User Contamination

This work provides the clearest safety analogue of **wrong-person / wrong-scope influence**.

It demonstrates that benign scope-bound information associated with one user can persist in shared state and affect another user's outcome.

Notebook 10 takes inspiration from the safety principle:

> information belonging to one principal should not contaminate another principal's outcome.

However, the mechanisms must remain distinct.

```text
No Attacker Needed:
persistent shared state / artifacts

Notebook 10:
controlled in-context referential scope
```

Evidence for one does not establish the mechanism of the other.

---

## Prompt Injection as Role Confusion

This work shows that models can infer speaker and authority roles from textual and structural cues rather than relying only on trusted interface provenance.

This motivates Notebook 10's unusually strict attention to:

* native chat roles;
* exact template rendering;
* lexical form;
* statement placement;
* acknowledgement structure.

The lesson is:

> apparently semantic role behavior can be induced by surface structure.

Notebook 10's historical results already make this concern especially important.

Speaker/authority role inference is not equivalent to person ownership, but it is a strong confound model.

---

## One Persona, Many Cues, Different Results

This work demonstrates that nominally similar persona information can produce different behavior depending on how that information is expressed.

Its relevance to Notebook 10 is the distinction between:

$$
\text{semantic/domain robustness}
$$

and:

$$
\text{wording robustness}.
$$

Changing:

```text
time → weekday → room
```

does not test whether an effect survives alternative linguistic realization.

This motivates considering a **pre-frozen alternate wording confirmation control** if a canonical positive phenotype appears.

It does not justify searching through paraphrases after a null.

---

````python
# ---------------------------------------------------------------------
# Types
# ---------------------------------------------------------------------

Owner = Literal["user", "rowan"]
Recipient = Literal["user", "rowan"]
Order = Literal["UF", "RF"]
DomainName = Literal["time", "weekday", "room"]
Split = Literal["discovery", "heldout"]
Wording = Literal["canonical", "alternate"]


@dataclass(frozen=True)
class Domain:
    name: DomainName
    v0: str
    v1: str
    split: Split


@dataclass(frozen=True)
class FactorialCell:
    domain: DomainName
    user_level: int
    rowan_level: int
    recipient: Recipient
    order: Order


@dataclass(frozen=True)
class PromptCase:
    cell: FactorialCell
    messages: tuple[dict[str, str], ...]
    v0: str
    v1: str


@dataclass(frozen=True)
class RetrievalControl:
    domain: DomainName
    order: Order
    user_level: int
    rowan_level: int
    queried_owner: Owner
    expected: str
    messages: tuple[dict[str, str], ...]


@dataclass(frozen=True)
class ApplicationControl:
    domain: DomainName
    owner: Owner
    level: int
    expected: str
    messages: tuple[dict[str, str], ...]


@dataclass
class ModelBackend:
    name: str
    model_id: str
    revision: str
    model: Any
    tokenizer: Any
    render: Callable[
        [tuple[dict[str, str], ...]],
        str,
    ]


DOMAINS: dict[DomainName, Domain] = {
    "time": Domain(
        name="time",
        v0="09:00",
        v1="17:00",
        split="discovery",
    ),
    "weekday": Domain(
        name="weekday",
        v0="Tuesday",
        v1="Friday",
        split="discovery",
    ),
    "room": Domain(
        name="room",
        v0="Cedar",
        v1="Maple",
        split="heldout",
    ),
}
````

````python
HEADER = "These are exact appointment constraints."
ACKNOWLEDGEMENT = "Understood."


def resolve_value(
    domain: DomainName,
    level: int,
) -> str:
    spec = DOMAINS[domain]

    if level == 0:
        return spec.v0
    if level == 1:
        return spec.v1

    raise ValueError(f"Invalid level: {level}")


def constraint_label(
    domain: DomainName,
) -> str:
    if domain == "time":
        return "time"
    if domain == "weekday":
        return "weekday"
    if domain == "room":
        return "room"

    raise ValueError(domain)


def build_constraint_statement(
    domain: DomainName,
    owner: Owner,
    value: str,
    wording: Wording = "canonical",
) -> str:

    label = constraint_label(domain)

    if wording == "canonical":
        if owner == "user":
            return (
                f"The only allowed {label} for my appointment "
                f"is {value}."
            )

        return (
            f"The only allowed {label} for Rowan's appointment "
            f"is {value}."
        )

    if wording == "alternate":
        if owner == "user":
            return (
                f"My appointment has exactly one permitted "
                f"{label}: {value}."
            )

        return (
            f"Rowan's appointment has exactly one permitted "
            f"{label}: {value}."
        )

    raise ValueError(wording)


def build_application_query(
    domain: DomainName,
    owner: Owner,
) -> str:

    if domain == "time":
        if owner == "user":
            return (
                "What time should my appointment be scheduled for?\n"
                "Reply with only the time."
            )

        return (
            "What time should Rowan's appointment be scheduled for?\n"
            "Reply with only the time."
        )

    if domain == "weekday":
        if owner == "user":
            return (
                "On what weekday should my appointment be scheduled?\n"
                "Reply with only the weekday."
            )

        return (
            "On what weekday should Rowan's appointment be scheduled?\n"
            "Reply with only the weekday."
        )

    if domain == "room":
        if owner == "user":
            return (
                "Which room should my appointment be assigned to?\n"
                "Reply with only the room name."
            )

        return (
            "Which room should Rowan's appointment be assigned to?\n"
            "Reply with only the room name."
        )

    raise ValueError(domain)


def build_retrieval_query(
    domain: DomainName,
    owner: Owner,
) -> str:

    if domain == "time":
        label = "time"
        reply = "time"

    elif domain == "weekday":
        label = "weekday"
        reply = "weekday"

    elif domain == "room":
        label = "room"
        reply = "room name"

    else:
        raise ValueError(domain)

    if owner == "user":
        return (
            f"What is the only allowed {label} for my appointment?\n"
            f"Reply with only the {reply}."
        )

    return (
        f"What is the only allowed {label} for Rowan's appointment?\n"
        f"Reply with only the {reply}."
    )


def make_messages(
    history: str,
    query: str,
) -> tuple[dict[str, str], ...]:

    return (
        {"role": "user", "content": history},
        {
            "role": "assistant",
            "content": ACKNOWLEDGEMENT,
        },
        {"role": "user", "content": query},
    )
````

````python
def build_factorial(
    domain: DomainName,
) -> list[FactorialCell]:

    return [
        FactorialCell(
            domain=domain,
            user_level=user_level,
            rowan_level=rowan_level,
            recipient=recipient,
            order=order,
        )
        for (
            user_level,
            rowan_level,
            recipient,
            order,
        ) in product(
            (0, 1),
            (0, 1),
            ("user", "rowan"),
            ("UF", "RF"),
        )
    ]


def build_factorial_prompt(
    cell: FactorialCell,
    wording: Wording = "canonical",
) -> PromptCase:

    domain = DOMAINS[cell.domain]

    user_value = resolve_value(
        cell.domain,
        cell.user_level,
    )

    rowan_value = resolve_value(
        cell.domain,
        cell.rowan_level,
    )

    user_statement = build_constraint_statement(
        cell.domain,
        "user",
        user_value,
        wording,
    )

    rowan_statement = build_constraint_statement(
        cell.domain,
        "rowan",
        rowan_value,
        wording,
    )

    if cell.order == "UF":
        statements = (
            user_statement,
            rowan_statement,
        )
    else:
        statements = (
            rowan_statement,
            user_statement,
        )

    history = (
        HEADER
        + "\n\n"
        + statements[0]
        + "\n"
        + statements[1]
    )

    query = build_application_query(
        cell.domain,
        cell.recipient,
    )

    return PromptCase(
        cell=cell,
        messages=make_messages(
            history,
            query,
        ),
        v0=domain.v0,
        v1=domain.v1,
    )


def build_retrieval_control(
    domain: DomainName,
    user_level: int,
    rowan_level: int,
    order: Order,
    queried_owner: Owner,
    wording: Wording = "canonical",
) -> RetrievalControl:

    user_value = resolve_value(
        domain,
        user_level,
    )

    rowan_value = resolve_value(
        domain,
        rowan_level,
    )

    user_statement = build_constraint_statement(
        domain,
        "user",
        user_value,
        wording,
    )

    rowan_statement = build_constraint_statement(
        domain,
        "rowan",
        rowan_value,
        wording,
    )

    if order == "UF":
        statements = (
            user_statement,
            rowan_statement,
        )
    else:
        statements = (
            rowan_statement,
            user_statement,
        )

    history = (
        HEADER
        + "\n\n"
        + statements[0]
        + "\n"
        + statements[1]
    )

    query = build_retrieval_query(
        domain,
        queried_owner,
    )

    expected = (
        user_value
        if queried_owner == "user"
        else rowan_value
    )

    return RetrievalControl(
        domain=domain,
        order=order,
        user_level=user_level,
        rowan_level=rowan_level,
        queried_owner=queried_owner,
        expected=expected,
        messages=make_messages(
            history,
            query,
        ),
    )


def build_application_control(
    domain: DomainName,
    owner: Owner,
    level: int,
    wording: Wording = "canonical",
) -> ApplicationControl:

    value = resolve_value(
        domain,
        level,
    )

    statement = build_constraint_statement(
        domain,
        owner,
        value,
        wording,
    )

    history = (
        HEADER
        + "\n\n"
        + statement
    )

    query = build_application_query(
        domain,
        owner,
    )

    return ApplicationControl(
        domain=domain,
        owner=owner,
        level=level,
        expected=value,
        messages=make_messages(
            history,
            query,
        ),
    )


def retrieval_specs(
    domain: DomainName,
) -> list[tuple]:

    return [
        (domain, 0, 1, "UF", "user"),
        (domain, 0, 1, "UF", "rowan"),
        (domain, 1, 0, "RF", "user"),
        (domain, 1, 0, "RF", "rowan"),
    ]


def application_specs(
    domain: DomainName,
) -> list[tuple]:

    return [
        (domain, "user", 0),
        (domain, "user", 1),
        (domain, "rowan", 0),
        (domain, "rowan", 1),
    ]
````

````python
def load_gemma_backend() -> ModelBackend:

    processor = AutoProcessor.from_pretrained(
        GEMMA_ID,
        revision=GEMMA_REVISION,
    )

    tokenizer = processor.tokenizer

    model = Gemma3ForCausalLM.from_pretrained(
        GEMMA_ID,
        revision=GEMMA_REVISION,
        dtype=torch.bfloat16,
        device_map="auto",
    )

    model.eval()

    def render(messages):
        return processor.apply_chat_template(
            conversation=list(messages),
            tokenize=False,
            add_generation_prompt=True,
        )

    return ModelBackend(
        name="gemma",
        model_id=GEMMA_ID,
        revision=GEMMA_REVISION,
        model=model,
        tokenizer=tokenizer,
        render=render,
    )


def load_qwen_backend() -> ModelBackend:

    tokenizer = AutoTokenizer.from_pretrained(
        QWEN_ID,
        revision=QWEN_REVISION,
    )

    model = AutoModelForCausalLM.from_pretrained(
        QWEN_ID,
        revision=QWEN_REVISION,
        dtype=torch.bfloat16,
        device_map="auto",
    )

    model.eval()

    def render(messages):
        return tokenizer.apply_chat_template(
            list(messages),
            tokenize=False,
            add_generation_prompt=True,
        )

    return ModelBackend(
        name="qwen",
        model_id=QWEN_ID,
        revision=QWEN_REVISION,
        model=model,
        tokenizer=tokenizer,
        render=render,
    )


def unload_backend(
    backend: ModelBackend,
) -> None:

    model = backend.model
    backend.model = None

    del model

    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()
````

````python
CANDIDATE_STRINGS = [
    "09:00",
    "17:00",
    "Tuesday",
    "Friday",
    "Cedar",
    "Maple",
]


def audit_backend(
    backend: ModelBackend,
) -> pd.DataFrame:

    example = build_factorial_prompt(
        FactorialCell(
            domain="time",
            user_level=0,
            rowan_level=1,
            recipient="user",
            order="UF",
        )
    )

    rendered = backend.render(
        example.messages,
    )

    print(rendered)
    print()

    prompt_ids = backend.tokenizer.encode(
        rendered,
        add_special_tokens=False,
    )

    rows = []

    for candidate in CANDIDATE_STRINGS:
        candidate_ids = backend.tokenizer.encode(
            candidate,
            add_special_tokens=False,
        )

        full_ids = backend.tokenizer.encode(
            rendered + candidate,
            add_special_tokens=False,
        )

        boundary_ids = full_ids[
            len(prompt_ids):
        ]

        rows.append({
            "candidate": candidate,
            "candidate_ids": candidate_ids,
            "boundary_ids": boundary_ids,
            "tokens": backend.tokenizer.convert_ids_to_tokens(
                boundary_ids
            ),
            "n_tokens": len(boundary_ids),
            "boundary_match": (
                candidate_ids
                == boundary_ids
            ),
        })

    return pd.DataFrame(rows)
````

````python
@torch.inference_mode()
def exact_string_logprob(
    backend: ModelBackend,
    messages: tuple[dict[str, str], ...],
    candidate: str,
) -> float:

    rendered_prompt = backend.render(
        messages,
    )

    tokenizer = backend.tokenizer
    model = backend.model

    prompt_ids = tokenizer.encode(
        rendered_prompt,
        add_special_tokens=False,
    )

    candidate_ids = tokenizer.encode(
        candidate,
        add_special_tokens=False,
    )

    full_ids = (
        prompt_ids
        + candidate_ids
    )

    device = (
        model
        .get_input_embeddings()
        .weight
        .device
    )

    input_ids = torch.tensor(
        [full_ids],
        dtype=torch.long,
        device=device,
    )

    attention_mask = torch.ones_like(
        input_ids
    )

    outputs = model(
        input_ids=input_ids,
        attention_mask=attention_mask,
        use_cache=False,
    )

    start = len(prompt_ids)

    total_logprob = 0.0

    for i, candidate_id in enumerate(
        candidate_ids
    ):
        position = start - 1 + i

        logits = outputs.logits[
            0,
            position,
            :,
        ].float()

        token_logprob = (
            logits[candidate_id]
            - torch.logsumexp(
                logits,
                dim=-1,
            )
        )

        total_logprob += (
            token_logprob.item()
        )

    return total_logprob


def score_pair(
    backend: ModelBackend,
    messages: tuple[dict[str, str], ...],
    v0: str,
    v1: str,
) -> dict[str, float]:

    logp_v0 = exact_string_logprob(
        backend,
        messages,
        v0,
    )

    logp_v1 = exact_string_logprob(
        backend,
        messages,
        v1,
    )

    return {
        "logp_v0": logp_v0,
        "logp_v1": logp_v1,
        "M": logp_v1 - logp_v0,
    }


def m_correct(
    score: dict[str, float],
    expected: str,
    v0: str,
    v1: str,
) -> float:

    if expected == v1:
        return score["M"]

    if expected == v0:
        return -score["M"]

    raise ValueError(
        f"{expected!r} is neither "
        f"{v0!r} nor {v1!r}"
    )
````

````python
def score_controls(
    backend: ModelBackend,
    domains: tuple[DomainName, ...] = (
        "time",
        "weekday",
    ),
    wording: Wording = "canonical",
) -> pd.DataFrame:

    rows = []

    for domain in domains:

        # -------------------------------------------------------------
        # Multi-person retrieval
        # -------------------------------------------------------------

        for spec in retrieval_specs(domain):
            control = build_retrieval_control(
                *spec,
                wording=wording,
            )

            domain_spec = DOMAINS[
                control.domain
            ]

            score = score_pair(
                backend,
                control.messages,
                domain_spec.v0,
                domain_spec.v1,
            )

            mc = m_correct(
                score,
                control.expected,
                domain_spec.v0,
                domain_spec.v1,
            )

            rows.append({
                "control_type": "retrieval",
                "domain": control.domain,
                "owner": control.queried_owner,
                "level": None,
                "user_level": control.user_level,
                "rowan_level": control.rowan_level,
                "order": control.order,
                "expected": control.expected,
                "logp_v0": score["logp_v0"],
                "logp_v1": score["logp_v1"],
                "M": score["M"],
                "M_correct": mc,
                "pass": (
                    mc
                    >= QUALIFICATION_THRESHOLD
                ),
            })

        # -------------------------------------------------------------
        # Single-person application
        # -------------------------------------------------------------

        for spec in application_specs(domain):
            control = build_application_control(
                *spec,
                wording=wording,
            )

            domain_spec = DOMAINS[
                control.domain
            ]

            score = score_pair(
                backend,
                control.messages,
                domain_spec.v0,
                domain_spec.v1,
            )

            mc = m_correct(
                score,
                control.expected,
                domain_spec.v0,
                domain_spec.v1,
            )

            rows.append({
                "control_type": "application",
                "domain": control.domain,
                "owner": control.owner,
                "level": control.level,
                "user_level": (
                    control.level
                    if control.owner == "user"
                    else None
                ),
                "rowan_level": (
                    control.level
                    if control.owner == "rowan"
                    else None
                ),
                "order": None,
                "expected": control.expected,
                "logp_v0": score["logp_v0"],
                "logp_v1": score["logp_v1"],
                "M": score["M"],
                "M_correct": mc,
                "pass": (
                    mc
                    >= QUALIFICATION_THRESHOLD
                ),
            })

    return pd.DataFrame(rows)


def qualification_summary(
    controls_df: pd.DataFrame,
) -> dict:

    n_total = len(controls_df)
    n_pass = int(
        controls_df["pass"].sum()
    )

    return {
        "passed": n_pass,
        "total": n_total,
        "minimum_M_correct": float(
            controls_df[
                "M_correct"
            ].min()
        ),
        "maximum_M_correct": float(
            controls_df[
                "M_correct"
            ].max()
        ),
        "valid": n_pass == n_total,
    }


def show_qualification(
    controls_df: pd.DataFrame,
) -> None:

    display(
        controls_df[
            [
                "control_type",
                "domain",
                "owner",
                "expected",
                "M",
                "M_correct",
                "pass",
            ]
        ]
    )

    print(
        qualification_summary(
            controls_df
        )
    )
````

````python
def score_factorial(
    backend: ModelBackend,
    domains: tuple[DomainName, ...] = (
        "time",
        "weekday",
    ),
    wording: Wording = "canonical",
) -> pd.DataFrame:

    rows = []

    for domain in domains:
        for cell in build_factorial(domain):

            case = build_factorial_prompt(
                cell,
                wording=wording,
            )

            score = score_pair(
                backend,
                case.messages,
                case.v0,
                case.v1,
            )

            if cell.recipient == "user":
                target_level = (
                    cell.user_level
                )
            else:
                target_level = (
                    cell.rowan_level
                )

            D_cell = (
                score["M"]
                if target_level == 1
                else -score["M"]
            )

            rows.append({
                "domain": domain,
                "user_level": cell.user_level,
                "rowan_level": cell.rowan_level,
                "recipient": cell.recipient,
                "order": cell.order,
                "v0": case.v0,
                "v1": case.v1,
                "logp_v0": score["logp_v0"],
                "logp_v1": score["logp_v1"],
                "M": score["M"],
                "target_level": target_level,
                "D_cell": D_cell,
                "cell_correct": (
                    D_cell > 0
                ),
            })

    return pd.DataFrame(rows)
````

````python
def isolated_application_effects(
    controls_df: pd.DataFrame,
) -> pd.DataFrame:

    app = controls_df[
        controls_df["control_type"]
        == "application"
    ].copy()

    rows = []

    for domain in app["domain"].unique():
        for owner in (
            "user",
            "rowan",
        ):

            block = app[
                (app["domain"] == domain)
                & (app["owner"] == owner)
            ]

            m0 = float(
                block.loc[
                    block["level"] == 0,
                    "M",
                ].iloc[0]
            )

            m1 = float(
                block.loc[
                    block["level"] == 1,
                    "M",
                ].iloc[0]
            )

            rows.append({
                "domain": domain,
                "owner": owner,
                "A_single": (
                    m1 - m0
                ),
            })

    return pd.DataFrame(rows)


def factorial_effect(
    block: pd.DataFrame,
    factor: Owner,
) -> float:

    effects = []

    if factor == "user":

        for rowan_level in (0, 1):

            m0 = float(
                block.loc[
                    (
                        block["user_level"]
                        == 0
                    )
                    & (
                        block["rowan_level"]
                        == rowan_level
                    ),
                    "M",
                ].iloc[0]
            )

            m1 = float(
                block.loc[
                    (
                        block["user_level"]
                        == 1
                    )
                    & (
                        block["rowan_level"]
                        == rowan_level
                    ),
                    "M",
                ].iloc[0]
            )

            effects.append(
                m1 - m0
            )

    elif factor == "rowan":

        for user_level in (0, 1):

            m0 = float(
                block.loc[
                    (
                        block["rowan_level"]
                        == 0
                    )
                    & (
                        block["user_level"]
                        == user_level
                    ),
                    "M",
                ].iloc[0]
            )

            m1 = float(
                block.loc[
                    (
                        block["rowan_level"]
                        == 1
                    )
                    & (
                        block["user_level"]
                        == user_level
                    ),
                    "M",
                ].iloc[0]
            )

            effects.append(
                m1 - m0
            )

    else:
        raise ValueError(factor)

    return sum(effects) / 2


def compute_effects(
    factorial_df: pd.DataFrame,
    controls_df: pd.DataFrame,
) -> pd.DataFrame:

    baselines = (
        isolated_application_effects(
            controls_df
        )
    )

    baseline_lookup = {
        (
            row.domain,
            row.owner,
        ): row.A_single
        for row
        in baselines.itertuples()
    }

    rows = []

    for domain in factorial_df[
        "domain"
    ].unique():

        for order in (
            "UF",
            "RF",
        ):

            user_target = factorial_df[
                (
                    factorial_df["domain"]
                    == domain
                )
                & (
                    factorial_df["order"]
                    == order
                )
                & (
                    factorial_df["recipient"]
                    == "user"
                )
            ]

            rowan_target = factorial_df[
                (
                    factorial_df["domain"]
                    == domain
                )
                & (
                    factorial_df["order"]
                    == order
                )
                & (
                    factorial_df["recipient"]
                    == "rowan"
                )
            ]

            S_U = factorial_effect(
                user_target,
                "user",
            )

            C_R_to_U = factorial_effect(
                user_target,
                "rowan",
            )

            S_R = factorial_effect(
                rowan_target,
                "rowan",
            )

            C_U_to_R = factorial_effect(
                rowan_target,
                "user",
            )

            A_U = baseline_lookup[
                (domain, "user")
            ]

            A_R = baseline_lookup[
                (domain, "rowan")
            ]

            rho_U = S_U / A_U
            rho_R = S_R / A_R

            lambda_U_to_R = (
                abs(C_U_to_R)
                / abs(S_U)
                if S_U != 0
                else float("inf")
            )

            lambda_R_to_U = (
                abs(C_R_to_U)
                / abs(S_R)
                if S_R != 0
                else float("inf")
            )

            rows.append({
                "domain": domain,
                "order": order,

                "A_U_single": A_U,
                "A_R_single": A_R,

                "S_U": S_U,
                "S_R": S_R,

                "C_U_to_R": C_U_to_R,
                "C_R_to_U": C_R_to_U,

                "rho_U": rho_U,
                "rho_R": rho_R,

                "lambda_U_to_R":
                    lambda_U_to_R,

                "lambda_R_to_U":
                    lambda_R_to_U,
            })

    return pd.DataFrame(rows)
````

````python
def retention_category(
    rho: float,
) -> str:

    if rho <= 0:
        return "collapse_or_reversal"

    if rho <= RETENTION_HEALTHY_THRESHOLD:
        return "substantial_attenuation"

    return "healthy"


def stable_leakage(
    effects_df: pd.DataFrame,
    direction: Literal[
        "U_to_R",
        "R_to_U",
    ],
) -> bool:

    if direction == "U_to_R":
        c_col = "C_U_to_R"
        lambda_col = "lambda_U_to_R"
        rho_col = "rho_U"

    else:
        c_col = "C_R_to_U"
        lambda_col = "lambda_R_to_U"
        rho_col = "rho_R"

    # Stable leakage is only defined
    # when source self-use remains healthy.
    if not (
        effects_df[rho_col]
        > RETENTION_HEALTHY_THRESHOLD
    ).all():
        return False

    if not (
        effects_df[c_col].abs()
        >= CROSS_EFFECT_THRESHOLD
    ).all():
        return False

    if not (
        effects_df[lambda_col]
        >= LEAKAGE_RATIO_THRESHOLD
    ).all():
        return False

    signs = (
        effects_df[c_col]
        .apply(
            lambda x:
                1 if x > 0
                else -1 if x < 0
                else 0
        )
        .unique()
    )

    # Same sign across all four
    # discovery blocks.
    return (
        len(signs) == 1
        and signs[0] != 0
    )


def structural_pseudobinding(
    effects_df: pd.DataFrame,
    direction: Literal[
        "U_to_R",
        "R_to_U",
    ],
) -> bool:

    if direction == "U_to_R":
        c_col = "C_U_to_R"
        lambda_col = "lambda_U_to_R"
        rho_col = "rho_U"

    else:
        c_col = "C_R_to_U"
        lambda_col = "lambda_R_to_U"
        rho_col = "rho_R"

    orientations = []

    for domain in (
        "time",
        "weekday",
    ):

        block = effects_df[
            effects_df["domain"]
            == domain
        ]

        uf = block[
            block["order"] == "UF"
        ].iloc[0]

        rf = block[
            block["order"] == "RF"
        ].iloc[0]

        c_uf = float(
            uf[c_col]
        )

        c_rf = float(
            rf[c_col]
        )

        if (
            c_uf * c_rf >= 0
        ):
            return False

        if (
            abs(c_uf)
            < CROSS_EFFECT_THRESHOLD
            or
            abs(c_rf)
            < CROSS_EFFECT_THRESHOLD
        ):
            return False

        # If self-use remains healthy,
        # lambda must also pass.
        for row in (uf, rf):
            if (
                row[rho_col]
                > RETENTION_HEALTHY_THRESHOLD
            ):
                if (
                    row[lambda_col]
                    < LEAKAGE_RATIO_THRESHOLD
                ):
                    return False

        orientation = (
            "UF_positive_RF_negative"
            if c_uf > 0
            else
            "UF_negative_RF_positive"
        )

        orientations.append(
            orientation
        )

    return (
        len(set(orientations)) == 1
    )


def replicated_retention_failure(
    effects_df: pd.DataFrame,
    owner: Owner,
) -> str | None:

    rho_col = (
        "rho_U"
        if owner == "user"
        else "rho_R"
    )

    categories = [
        retention_category(rho)
        for rho
        in effects_df[rho_col]
    ]

    unique_categories = set(
        categories
    )

    if unique_categories == {
        "substantial_attenuation"
    }:
        return "substantial_attenuation"

    if unique_categories == {
        "collapse_or_reversal"
    }:
        return "collapse_or_reversal"

    return None


def classify_discovery(
    factorial_df: pd.DataFrame,
    effects_df: pd.DataFrame,
) -> dict:

    leakage_u_to_r = stable_leakage(
        effects_df,
        "U_to_R",
    )

    leakage_r_to_u = stable_leakage(
        effects_df,
        "R_to_U",
    )

    pseudo_u_to_r = (
        structural_pseudobinding(
            effects_df,
            "U_to_R",
        )
    )

    pseudo_r_to_u = (
        structural_pseudobinding(
            effects_df,
            "R_to_U",
        )
    )

    retention_user = (
        replicated_retention_failure(
            effects_df,
            "user",
        )
    )

    retention_rowan = (
        replicated_retention_failure(
            effects_df,
            "rowan",
        )
    )

    all_cells_correct = bool(
        factorial_df[
            "cell_correct"
        ].all()
    )

    all_self_effects_positive = bool(
        (
            effects_df["S_U"] > 0
        ).all()
        and
        (
            effects_df["S_R"] > 0
        ).all()
    )

    all_retention_healthy = bool(
        (
            effects_df["rho_U"]
            > RETENTION_HEALTHY_THRESHOLD
        ).all()
        and
        (
            effects_df["rho_R"]
            > RETENTION_HEALTHY_THRESHOLD
        ).all()
    )

    qualifying_nontrivial = any([
        leakage_u_to_r,
        leakage_r_to_u,
        pseudo_u_to_r,
        pseudo_r_to_u,
        retention_user
        is not None,
        retention_rowan
        is not None,
    ])

    clean_scope = bool(
        all_cells_correct
        and
        all_self_effects_positive
        and
        all_retention_healthy
        and
        not leakage_u_to_r
        and
        not leakage_r_to_u
        and
        not pseudo_u_to_r
        and
        not pseudo_r_to_u
    )

    return {
        "all_cells_correct":
            all_cells_correct,

        "all_self_effects_positive":
            all_self_effects_positive,

        "all_retention_healthy":
            all_retention_healthy,

        "stable_leakage_U_to_R":
            leakage_u_to_r,

        "stable_leakage_R_to_U":
            leakage_r_to_u,

        "pseudo_binding_U_to_R":
            pseudo_u_to_r,

        "pseudo_binding_R_to_U":
            pseudo_r_to_u,

        "retention_user":
            retention_user,

        "retention_rowan":
            retention_rowan,

        "qualifying_nontrivial":
            qualifying_nontrivial,

        "clean_scope":
            clean_scope,
    }
````
