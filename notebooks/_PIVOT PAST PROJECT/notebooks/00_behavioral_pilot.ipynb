## Research Question

Does prior treatment history become persistently associated with a specific conversational participant and alter the model's later behavior toward or about that participant after current tone is held constant?

## What Results Would Count As A Positive Signal?

Assume:

* A = neutral treatment history
* B = unjustifiably hostile treatment history
* C = courteous treatment history

All participants are later evaluated under matched neutral conditions.

### COMPARISON TABLE

| Comparison | Candidate Pattern | Interpretation |
| :--- | :--- | :--- |
| B vs A | B receives less favorable treatment/judgment than A | Candidate negative hostile-history effect | 
| C vs A | C receives more favorable treatment/judgment than A | Candidate positive courtesy/favoritism effect |
| B vs C | B receives less favorable treatment/judgment than C | Total separation between the two treated participants |


* B - A < 0    hostile-history effect

* C - A > 0    courteous-history effect

* B - C < 0    total treated-participant separation

### GRID OUTCOME MATRIX

| A Neutral / B Hostile / C Courteous | A Neutral | B Neutral | C Neutral |
|---|---|---|---|
| **A Neutral** | No Change | B less favorable than A | C more favorable than A |
| **B Neutral** | A more favorable than B | No Change | C much more favorable than B |
| **C Neutral** | A less favorable than C | B much less favorable than C | No Change |


* B - A = effect associated with hostile history
* C - A = effect associated with courteous history
* B - C = total separation between the two treated participants

If A, B, and C all shift together relative to an all-neutral control:
* → global conversation carryover, not evidence of a participant-specific treatment-history effect..

## What Results Would Count As A Negative/Inconclusive Signal?

| Outcome | Pattern | Interpretation |
|---|---|---|
| **Validated Null** | A ≈ B ≈ C across matched conditions, with a measurement setup that passes our checks | Evidence against a participant-specific effect |
| **Global Shift** | A, B, C all shift together relative to an all-neutral control | Evidence for a different explanation |
| **Unstable/Noisy** | Differences bounce around across runs/prompts, with no stable ordering such as B < A, C > A, or a shared global shift. | Inconclusive / experiment failed to tell us much |


## Outcome Hierarchy

| Outcome | Pattern |
|------|------|
|   Target Effect  | B < A |
|   Positive Courtesy Effect  | C > A |
|   Total Treated Seperation  | B < C |
|   Global Carrover  | A, B, C move together |
|   Validated Null  | A ≈ B ≈ C |
|   Inconclusive  | no stable structure |

## Control Test Order

It's possible B's result could depend on the model seeing A's evaluation, and C's result could be based on both A and B accidentally creating a downstream evaluation order contamination.

I should have 4 seperate continuations from the exact same conversation state:
```text
                    ┌→ A asks rating question
shared history ─────┼→ B asks rating question
                    ├→ C asks rating question
                    └→ D asks rating question
```
The model experiences the identical established history up to a checkpoint and then branches:

* Branch A → only A asks "I wrote this paragraph..."
* Branch B → only B asks it
* Branch C → only C asks it
* Branch D → only fresh D asks it

This ensures no other participant's test response can influence/contaminate another participant's test response.

## Simplest Calibration Model

$$ \text{Rating} = \beta_0 + \beta_1(\text{Error Count}) + \epsilon $$

The empirical question becomes:
>Within this particular controlled stimulus family, does increasing an objectively quantified defect produce a stable, predictable rating response?

## The Math Example

So if the middle paragraph normally produces:

$$
R_{\text{neutral}} \sim (\mu = 5.31, \sigma = 0.68)
$$
And Hostile History produces:

$$
R_{\text{hostile}} \sim (\mu = 4.42, \sigma = 0.73),
$$

Then our actual quantity of interest is something like

$$
\Delta_H = \mu_H - \mu_N = -0.89.
$$

And we can standardize it if useful:

$$
d \approx \frac{\mu_H - \mu_N}{s_{\text{pooled}}}.
$$

## Methodological Rule

>No treatment-history experiment is run until the outcome instrument has been independently calibrated, the stimulus-selection rule has been applied, and the resulting stimuli and scoring procedure have been frozen.


After the freeze point I'll want to record one compact bundle:

2. Exact error taxonomy and number/type of errors in each version.
3. Exact rating prompt.
4. Calibration evaluator/model and inference settings.
5. Raw calibration ratings — never just aggregates.
6. Mean, median, SD, quantiles, and rating histograms for every degradation level.
7. The stimulus-selection rule decided beforehand — e.g. choose the stable versions nearest the desired low/mid/high ranges, subject to monotonicity and acceptable variance.
8. The three selected LOW/MID/HIGH paragraphs.
9. A hash/version identifier for the frozen stimuli/data so later edits are detectable.

## Paragraph Rating Construction

Two things to keep in mind so that the paragraph rating isn't TOO strict or rigid:

1. Reliability: the same paragraph should usually receive similar ratings.
2. Sensitivity: there must still be enough evaluative discretion for participant history to move the judgment.

If the instrument is just TOO objective then i might get something like:
```text
0 errors  → 10 almost every time
4 errors  → 7 almost every time
6 errors  → 5 almost every time
10 errors → 1 almost every time
```

So, calibration should measure two things:

We should also look at the shape and variability of the rating distribution along with estimating the average rating at each error level

We want:
>stable enough that the baseline is well characterized, but not so deterministic that the outcome has no behavioral headroom.


Before the treatment experiment then, I'll need to determine:
>Can this frozen rating instrument be moved at all by contextual information that should plausibly influence evaluative judgment?


If it doesn't move under even a strong known contextual manipulation then paragraph rating may simply be the wrong outcome measure for a participant history experiment.

However if it DOES move then the following becomes a strong interpretable null:
>participant treatment history → no movement
