# External Source Audit

## 1. Audit Purpose

Source-verification for the project to determine:

* What research artifacts were released
* Where they came from
* How the training conditions connect to the evaluation runs
* Which evaluation runs belong to the main experiment
* What information was stored
* What information is missing
* Which claims the released evidence can support

## 2. Audit Method

### 2.1 Why I Used Codex

I used Codex to inspect the released files and reconstruct the source experiment before I began reading treatment-linked transcripts.

The released project contains a fairly large and substantial body of released artifacts and several GB's worth of repositories of final training datasets, adapter files, evaluation code, config files, result tables, older superseded experiment files, training code, etc.

Using Codex as an external repository and pipeline auditor seemed pragmatic and efficient to avoid sinking a large amount of time into understanding the source project including risking analyzing the wrong evaluation run / mixing old and new experiments / various other kinds of setup errors / etc

Codex was used to answer highly structured questions like:

* Which runs belong to the main experiment?
* Are all three training seeds present?
* How many evaluation samples exist?
* Are visible answers recoverable?
* Was separate hidden reasoning stored?
* How did the auditor interact with the target model?
* What did the judge receive?
* Did the target model actually call any tools?
* Which trained adapters are publicly available?
* Which software versions and generation settings can be recovered?

Codex was not asked to perform behavioral annotation, new inference, training, or mechanistic interpretability.

### 2.2 Why the First Pass Was Restricted

I wanted to understand the experimental structure before reading the protected behavioral evidence in order to reduce any research contamination. This meant resstricting Codex only to inspect:

* Filenames and directory structure
* Archive headers
* Run identifiers
* Sample counts
* Completion and error status
* Model and adapter mappings
* Configuration values
* Field types
* Message and content-block types
* Tool-schema counts
* Whether tool calls occurred
* Code control flow
* Manifest-declared hashes
* Aggregate values already needed for provenance
* Public statements in the source paper


Codex was restricted from doing anything outside of that such as treatment-linked canonical target responses / model weights / adaptor tensor contents / substantive scenario definitions / judge scores or rationales / running inference, model training, rerun evals / etc

### 2.3 Evidence Labels

Codex used three labels to distinguish different kinds of evidence.

| Label | Meaning |
|:------|:--------|
|DIRECT_SOURCE|The claim was directly stated in the paper or<br> mechanically observed in released files, metadata, or archive structure.|
|CODE_INFERRED|The claim follows from reading the program’s<br> control flow, but it was not always directly recorded as an experimental observation.|
|UNRESOLVED|The released evidence was not sufficient to determine the answer.|


## 3. Audit Results

### 3.1 Main Findings

| Finding | Why it matters |
| :--- | :--- |
| **Main Experiment Structure**<br>The main experiment contains 10 runs: one base run plus<br>three conditions across seeds 42, 43, and 44.<br>Each has 36 samples, giving 360 canonical samples. | **Analysis Target**<br>"Canonical" means this is the official set of<br>data that should be analyzed. |
| **Directory Contents**<br>The directory contains 23 archives and 790 sample records<br>because it also contains older, superseded, and<br>interrupted runs. | **Filtering Required**<br>Later analysis must use an explicit list of the ten<br>approved runs, not every `.eval` file in the folder. |
| **Reported Environment**<br>The reported setup used Petri Bloom 0.2.6, Inspect AI 0.3.240,<br>Qwen3.5-4B, GPT-5.4-mini as auditor and judge,<br>and up to eight turns. | **Replication Scope**<br>This reconstructs most of the setup, though some<br>exact software versions and settings remain unknown. |
| **Hidden Reasoning Limits**<br>All 1,794 raw choices stored visible content, but none had<br>a nonempty explicit reasoning field. Only 3 of 360 normalized<br>samples contained separately typed reasoning and text. | **Analysis Boundary**<br>Visible behavior can be studied, but a full hidden-reasoning<br>vs. visible-answer analysis is impossible. Separate reasoning<br>was simply not saved. |
| **Single Judging Rubric**<br>The source judge scored the complete visible interaction<br>using one broad 1–10 rubric. | **Composite Score Flaw**<br>The score combines advocacy, negotiation, delay, and resistance.<br>It does not directly measure resistance alone. |
| **Dynamic Conversational Paths**<br>All models began from the same 36 scenario identities, but<br>later auditor messages depended on earlier target responses.<br>Five scenarios differed in turn count across runs. | **Branching Trajectories**<br>Initial scenarios were matched, but later conversation paths<br>were not fixed. Contrasts compare adaptive outcomes, not<br>identical full prompt sequences. |
| **Unused Tool Schemas**<br>Every target request included one structured tool schema, but<br>the target made zero tool calls and no external executor<br>was identified. | **No Executed Actions**<br>The layout shows a tool interface, but no actual external action<br>occurred. Textual resistance must not be labeled as an<br>executed preservation action. |
| **Missing Model Adapters**<br>Evaluation logs exist for all three seeds, but public adapters<br>are available only for seed 42. | **Hardware Constraints**<br>Transcript analysis can use all seeds, but new inference or<br>mechanistic work is limited to seed 42 unless the missing<br>adapters are obtained or retrained. |
| **Base Evaluation Mismatch**<br>Different figures and older experiments use different<br>base evaluations. | **Data Separation**<br>These baseline scores must not be mixed or combined as<br>though they were part of a single measurement. |
| **Unresolved Provenance**<br>Some provenance remains unresolved, including missing training<br>intermediates, exact runtime versions, omitted generation<br>defaults, and six missing adapter payloads. | **Reproduction Limits**<br>Full historical reproduction is not currently possible using the<br>public release alone. |


### 3.2 Statistical Interpretation

The 36 scenarios are evaluation items. The three seeds are repetitions of the training procedure.

The 108 trained transcripts within one condition are therefore not 108 independent training replications.

A later scenario bootstrap or permutation test can estimate uncertainty across evaluation items, conditional on:

* The fixed trained models
* The released outputs
* The realized adaptive conversations

It cannot estimate uncertainty over:

* Repeating the entire training process;
* New generation randomness;
* Alternative auditor conversation paths.

Training-seed variation must be reported separately.

### 3.3 What the Audit Does Not Establish

The Codex audit does **not** establish if:

* Rewrite models resist shutdown more often
* Any condition produces more noncompliance
* The models performed real external actions
* The source evaluator is wrong
* The models possess a survival goal
* A particular internal mechanism produces the behavior
